```python
# EnumKraft – Dunkelflaute exploration
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print("EnumKraft environment ready")
```

```python
# Synthetic data – placeholder for real ENTSO-E
dates = pd.date_range("2024-01-01", "2024-12-31", freq="D")
wind_solar = 100 * np.exp(-((dates - dates[180]).days / 90)**2) + np.random.normal(0, 8, len(dates))
wind_solar = np.maximum(wind_solar, 5)

plt.figure(figsize=(12,4))
plt.plot(dates, wind_solar)
plt.title("Dunkelflaute test pattern (synthetic)")
plt.ylabel("Wind+Solar (a.u.)")
plt.grid(True)
plt.show()
```
